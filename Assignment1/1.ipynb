```python
import torch

# Create a tensor for nitrogen rate with gradient tracking
f = torch.tensor(80.0, requires_grad=True)

# Define the yield gain function: Y(f) = 24f- 0.1f^2
Y = 24 * f - 0.1 * f**2

# Compute the derivative using autograd
Y.backward()

# The gradient is stored in f.grad
print(f"Nitrogen rate: {f.item()} kg/ha")
print(f"Yield gain: {Y.item()} kg/ha")
print(f"Derivative Y'(80): {f.grad.item()}")
```

```python
import torch

rates = [40.0, 80.0, 120.0, 160.0, 200.0]

print("N rate | Yield gain | Y'(f) Autograd | Y'(f) Analytical")
print("-" * 60)

for rate_val in rates:

    # Create tensor with gradient tracking
    f = torch.tensor(rate_val, requires_grad=True)

    # Yield gain function
    Y = 24*f - 0.1*f**2

    # Compute gradient
    Y.backward()

    # Analytical derivative: Y'(f) = 24-0.2f
    analytical = 24 - 0.2*rate_val

    print(f"{rate_val:6.0f} | {Y.item():10.0f} |"
          f" {f.grad.item():14.1f} | {analytical:14.1f}")
```
